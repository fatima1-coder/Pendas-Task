```python
import pandas as pd
```

```python
#Task 1
data={
    "Name":["Ali","Sara","Aisha","Zain","Ahmed"],
    "Marks":[70,85,95,45,60]

}
df=pd.DataFrame(data)
print(df)
marks=df["Marks"]
print(marks)
print(marks.mean())
print(marks.min())
print(marks.max())


total_students=len(df)
print(total_students)
```

```python
#Task 2
data={
    "Name":["Ali","Sara","Aisha","Zain","Ahmed"],
    "Marks":[70,85,95,45,60]

}
df=pd.DataFrame(data)
print(df)
passed=df[df["Marks"]>50]
print(passed)
failed=df[df["Marks"]<50]
print(failed)
```

```python
#Task 3
data={
    "Name":["Ali","Sara","Aisha","Zain","Ahmed"],
    "Marks":[70,85,95,45,60]

}
df=pd.DataFrame(data)
print(df)
df["city"]=["Karachi","Lahore","Karachi","Multan","islamabad"]
print(df)
karachites=df[df["city"]=="Karachi"]
print(karachites)
```

```python
#Task 4
data={
    "Name":["Hamza","Zara","Saba","Hashir","Ahad"],
    "Course":["AI","Web","AI","Web","AI"],
    "Marks":[90,46,75,45,60]

}
df=pd.DataFrame(data)
print(df)
result = df.groupby("Course")["Marks"].mean()
print(result)
```

```python
#Task 5
data={
    "Name":["Ali","Sara","Aisha","Zain","Ahmed"],
    "Marks":[80,None,None,90,70]

}
df=pd.DataFrame(data)
print(df)
print(df.isnull().sum())
df["Marks"]=df["Marks"].fillna(df["Marks"].mean())
print(df["Marks"])
df=df.dropna
print(df)
```
